#### b) Creating a new column - Age Group

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['Age Group'] = pd.qcut(df['Age'], q=4, labels = labels)
```

q=4: Tells pandas to split the *Age data* into 4 equal-sized buckets (quartiles: 0–25%, 25–50%, 50–75%, and 75–100% of the dataset). 
This ensures that roughly the same number of rows fall into each age group.

#### b) Query 1: Total Revenue and Average Spend by Product Category

Business Objective: Identify which product categories drive the highest revenue

```python
import pandas as pd
query_category = """
SELECT 
    Category,
    COUNT("Customer ID") AS Total_Transactions,
    SUM("Purchase Amount (USD)") AS Total_Revenue,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Average_Spend,
    ROUND(AVG("Review Rating"), 2) AS Average_Rating
FROM customer_transactions

GROUP BY Category
ORDER BY Total_Revenue DESC;
"""

category_result = pd.read_sql(query_category, conn)
category_result
```

Finding: **Clothing** and **Accessories** are the primary revenue drivers for the 
retail company, contributing significantly more sales volume than Footwear
and Outerwear, while average purchase value remains consistent across all 
categories 57 to 60 dollar.

#### c) Query 2: High-Value Customers by Subscription Status

```python
query_segments = """
SELECT 
    "Subscription Status",
    "Discount Applied",
    COUNT("Customer ID") AS Total_Customers,
    ROUND(AVG("Previous Purchases"), 1) AS Avg_Previous_Purchases,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Avg_Purchase_Amount
FROM customer_transactions

GROUP BY "Subscription Status", "Discount Applied"
ORDER BY Avg_Purchase_Amount DESC;
"""

segment_result = pd.read_sql(query_segments, conn)
segment_result
```

Consistent Spending: The average purchase amount remains remarkably 
stable (58.92 to 60.13) across all combinations, **showing that discounts 
and subscriptions do not heavily distort individual transaction sizes.**
    
Non-Subscriber Base: The **largest segment is non-subscribers who buy without 
discounts (2,223 customers)**, representing a massive target market for 
future loyalty program upsells.

#### d)  Regional Performance: Top Locations by Revenue
Management wants to know which geographic locations yield the highest total sales and how ratings compare across regions.


```python
query_top_locations = """
SELECT 
    Location,
    COUNT("Customer ID") AS Total_Transactions,
    SUM("Purchase Amount (USD)") AS Total_Revenue,
    ROUND(AVG("Purchase Amount (USD)"), 2) AS Avg_Spend,
    ROUND(AVG("Review Rating"), 2) AS Avg_Rating
FROM customer_transactions

GROUP BY Location
ORDER BY Total_Revenue DESC
LIMIT 5;
"""

top_locations_df = pd.read_sql(query_top_locations, conn)
top_locations_df
```

**Montana** leads all regions with the highest total revenue (5,784) and transaction volume (96), followed closely by Illinois and California.

**Nevada** stands out with the highest average spend per order (63.38) despite having fewer total transactions (87), indicating higher basket sizes in that region.

#### e) Multi-Column Grouping: Payment Method & Shipping Type Combinations
To optimize checkout experiences and logistics, management wants to see which payment preferences are most frequently paired with specific shipping methods.

```python
query_payment_shipping = """
SELECT 
    "Payment Method",
    "Shipping Type",
    COUNT("Customer ID") AS Transactions,
    SUM("Purchase Amount (USD)") AS Total_Revenue
FROM customer_transactions
GROUP BY "Payment Method", "Shipping Type"
ORDER BY Total_Revenue DESC
LIMIT 5;
"""

combo_df = pd.read_sql(query_payment_shipping, conn)
combo_df
```

#### g) Using Spending Tier in SQL (Behavioral Value Segmentation)
Because we categorized spending into tiers (Low, Medium, High), you can filter or group by Spending Tier to see how subscription status or discounts correlate with spending level:


```python
query_spending_tier = """
SELECT 
    "Spending Tier",
    "Subscription Status",
    COUNT("Customer ID") AS Transaction_Count,
    SUM("Purchase Amount (USD)") AS Total_Revenue
FROM customer_transactions
GROUP BY "Spending Tier", "Subscription Status"
ORDER BY "Spending Tier", Total_Revenue DESC;
"""

spending_df = pd.read_sql(query_spending_tier, conn)
spending_df
```

Highlights a major growth opportunity: introducing loyalty perks 
to convert **non-subscribers** (947 in the high tier) into recurring subscribers. Aso medium tier has count of 953 who have not subscribed.

#### h) Using Frequency of Purchases in SQL (Loyalty & Retention)
To find out how often customers return and which frequency category generates the most dependable revenue, query the frequency column directly:


```python
query_frequency = """
SELECT 
    "Frequency of Purchases",
    "Purchase Frequency Days",
    COUNT("Customer ID") AS Total_Customers,
    ROUND(AVG("Previous Purchases"), 1) AS Avg_Previous_Purchases,
    SUM("Purchase Amount (USD)") AS Total_Revenue
FROM customer_transactions
GROUP BY "Frequency of Purchases"
ORDER BY Total_Revenue DESC;
"""

frq_df = pd.read_sql(query_frequency, conn)
frq_df
```
